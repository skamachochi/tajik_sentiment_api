```python
import numpy as np
import pandas as pd
import os
import re
```

```python
def clean_tajik(text):
    tajik_pattern = r'[^а-яА-ЯёЁғҒӣӢқҚӯҮҳҲҷҶ\s]'
    text = re.sub(tajik_pattern, '', text)
    text = text.strip()
    return text
```

```python
file_path = r"/Users/azamat/Desktop/sentiment API/dataset/train_data.csv"
df = pd.read_csv(file_path)
df['Comment'] = df['Comment'].apply(clean_tajik)
df.to_csv(file_path, index=False)
df.head()
```
